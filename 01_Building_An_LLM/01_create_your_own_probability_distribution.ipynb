```python
import random
```

### A) An example with all needed libs
#### 1) Setting a word probability example

```python
# First words and probabilities : 

def test_word_probabilities(word_probabilities: dict[str, float]) -> None: 
    # First check it is a dictionnary
    assert isinstance(word_probabilities, dict), f"Expected dict, got {type(word_probabilities)}"
    # The check each key-value pair
    sum_probs= 0.0
    for word, prob in word_probabilities.items():
        # Check key is a string and value is a float between 0 and 1
        assert isinstance(word, str), f"Expected str key, got {type(word)}"
        assert isinstance(prob, float), f"Expected float value, got {type(prob)}"
        assert 0.0 <= prob <= 1.0, f"Probability {prob} for word '{word}' is out of bounds"
        # Sum probabilities
        sum_probs += prob
    # Check that the sum of probabilities is approximately 1
    assert abs(sum_probs - 1.0) < 1e-6, f"Sum probs is {sum_probs}" # Tolerance for floating point sum
    print("All tests passed! You've set up the word probabilities correctly.")
    

word_probabilities = { # Not optimal for large dicts but just to have a good structure and see each word wrt its probability
    "star": 0.99,
    "beef": 0.001,
    "computer": 0.005,
    "ocean": 0.003,
    "mountain": 0.0005,
    "forest": 0.0005
}

test_word_probabilities(word_probabilities)
```

#### 2) Choosing random words with random choice

```python
choose_word = random.choices(
    population=list(word_probabilities.keys()),
    weights=list(word_probabilities.values()), 
    k=1
)

print(f"Twinkle twinkle little {choose_word[0]}!")
```

### B) Actual Exercise : 

```python
def prompt_generator(chosen_word: str) -> str:
    return f"Jide was hungry so she went looking for {chosen_word}!"

new_word_probabilites = {
    "food" : 0.32, 
    "snacks": 0.28,
    "leftovers": 0.15,
    "water": 0.10,
    "for" : 0.08,
    "her": 0.05, 
    "photosynthesis":0.01, 
    "pyramids": 0.005,
    "car": 0.002,
    "rocks": 0.002,
    "orthodontist": 0.001}

test_word_probabilities(new_word_probabilites)
```

```python
chosen_word = random.choices(
    population=list(new_word_probabilites.keys()),
    weights=list(new_word_probabilites.values()), 
    k=1
)[0]
print(prompt_generator(chosen_word))
```

An LLM produces multiple probabilities when prompted a set of words (for the next word to come). Such simplified way of showing how it would complete the sentence is the applied method in LLMS. (We will learn also about top k and top p to trunk list of probabilities for new_words). 

Another aspect to see is the probabilities by themselves. They are trained from the data. If the data is highly skewed towards a specific side (stereotypes ...), the model is affected. The probability of hallucinations by AI is not 0 when trained on not pruned data for validity (which would have required huge human effort in processing and cleaning a lot of content + control of information origins). 

A quick information to link with the next section is that generally the probabilities that we have are conditional !
