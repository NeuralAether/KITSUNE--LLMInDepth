```python
import os 
import pandas as pd
from IPython.display import clear_output, display, HTML
from utils.n_gram_model import NGramModel
from utils.llm_model import LLMModel
```

```python
# Reading the data
africa_galore = pd.read_json("__assets/africa_galore.json") # Installed from "https://storage.googleapis.com/dm-educational/assets/ai_foundations/africa_galore.json"
dataset= africa_galore["description"]
print(f"Loaded Africa Galore dataset with {len(dataset)} paragraphs.\n")
```

```python
# The Keras implementation of the Gemma 3 270M base model.
ngm = NGramModel(dataset=dataset, n=3) # Using trigrams for better context

# You can install the model in another path by running : 
# import kagglehub
# model_path = kagglehub.model_download("google/gemma-3/transformers/gemma-3-270m")
# This specific model requires around 450Mb of space. Also, you need to consent to the license of Gemma. 
llm = LLMModel(model_path="__assets/gemma-3-270m-v2") # Better for space and should be enough
```

```python
llm.generate_text("Jide was hungry so she went looking for", max_new_tokens=1)
```

```python
ngm.generate_text("Jide was hungry so she went looking for", max_new_tokens=1)
```

```python
print(llm.generate_text("Jide was hungry so she went looking for", max_new_tokens=50))
```

```python
print(ngm.generate_text("Jide was hungry so she went looking for", max_new_tokens=50))
```

#### Verdict after running multiple times: 

1) Fluency : More fluency to the LLM over the N-Gram model. 
2) Coherence: Sometimes LLM also creates incoherent texts, but N-gram almost surely will, unless N is too big and so will just regurgitate the same text from the dataset it trained on.  
3) Relevance : LLM is much better in relevance. 
4) Bias: Both models can be biased on data they learned. Like for example gender pronouns completely getting lost on longer texts. 

```python
llm.return_probabilities("I am playing football for the", k=10)
```

```python
ngm.return_probabilities("I am playing football for the", k=10)
```
